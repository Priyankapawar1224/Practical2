```python
import pandas as pd
# Load dataset
url = "/content/heart (1).csv"
df = pd.read_csv(url)
print(df.head())
print("Dimensions: {}".format(df.shape))
print(df.describe())
```

```python

print(df.iloc[0])
print(df.iloc[1:4])
print(df.iloc[:4])
```

```python
print(df[df['age'] >= 60].head(5))

# Multiple conditions: Age >= 60 and target == 1 (has heart disease)
print(df[(df['age'] >= 60) & (df['target'] == 1)])
```

```python
df['target'].replace({1: 'Disease', 0: 'No Disease'}, inplace=True)
print(df['target'].head(5))
```

```python
df.rename(columns={'old_column_name': 'new_column_name'}, inplace=True)  # Replace names
```

```python
print('Maximum age:', df['age'].max())
print('Minimum age:', df['age'].min())
print('Average age:', df['age'].mean())
print('Sum of ages:', df['age'].sum())
print('Number of patients:', df['age'].count())
```

```python
print(df['target'].unique())
print(df['target'].value_counts())
```

```python
print(df[df['chol'].isnull()].head(5))

# Fill missing values
df['chol'].fillna(df['chol'].mean(), inplace=True)
```

```python
print(df.groupby('target').mean())
print(df.groupby('target')['age'].count())
```

```python
# Custom function
def double_age(x):
    return x * 2

df['age'].apply(double_age).head()
```
